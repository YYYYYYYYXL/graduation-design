```python
import sys
from pathlib import Path

current = Path.cwd()
project_root = current.parent

sys.path.insert(0, str(project_root))


from src.datasets.sidd_dataset import SIDDDataset
from torch.utils.data import DataLoader
```

```python
train_dataset = SIDDDataset(
    split="train",
    patch_size=128,
    patches_per_image=4
)

val_dataset = SIDDDataset(split="val")
test_dataset = SIDDDataset(split="test")

print("Train:", len(train_dataset))
print("Val:", len(val_dataset))
print("Test:", len(test_dataset))
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=8,
    shuffle=True,
    num_workers=0
)

noisy_batch, gt_batch = next(iter(train_loader))

print("NOISY:", noisy_batch.shape)
print("GT:", gt_batch.shape)
print("数据类型:", noisy_batch.dtype)
print(f"像素范围:, {noisy_batch.min().item()} ----- {noisy_batch.max().item()}")
```
